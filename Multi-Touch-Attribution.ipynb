```python
# 1. Reload raw data fresh
df = pd.read_csv('C:/Users/sakin_xyc6lmd/OneDrive/Desktop/Infotact internship projects/Multi-Touch_Attribution_Project/multi-touch-attribution/multi_touch_attribution_data.csv')

# 2. Clean it
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
df['Campaign'] = df['Campaign'].replace('-', 'No Campaign')
df['Conversion'] = df['Conversion'].map({'Yes': 1, 'No': 0})

# 3. Rename and load
df_sql = df.rename(columns={
    'User ID': 'user_id',
    'Timestamp': 'timestamp',
    'Channel': 'channel',
    'Campaign': 'campaign',
    'Conversion': 'conversion'
})

df_sql.to_sql("interactions", con=engine, if_exists="replace", index=False)
print("Loaded", len(df_sql), "rows into interactions")
```
